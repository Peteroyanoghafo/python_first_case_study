#### Daily Weather Observation <br>
The field officer provides the following information. <br>
Information:
Value <br>
Farm Location:
Kaduna <br>
Crop Type:
Maize <br>
Temperature (°C):
31 <br>
Rainfall (mm):
18 <br>
Humidity (%):
72 <br>
Wind Speed (km/h):
16 <br>
Minimum Rainfall Required (mm):
20 <br>
Maximum Safe Temperature (°C):
35 <br>
Irrigation System Available:
True <br>




```python
farm_location = "Kaduna"
crop_type = "Maize"
temperature_C = 31
rainfall = 18
humidity = 72
wind_speed = 16
min_rainfall_required = 20
max_safe_temp_C = 35
irrigation_system_available = True

rainfall_difference = min_rainfall_required - rainfall
temperature_margin = max_safe_temp_C - temperature_C
rainfall += 8
rainfall_requirement_met = rainfall >= min_rainfall_required
temperature_safe = temperature_C <= max_safe_temp_C
cropping_favored = rainfall_requirement_met and temperature_safe
farmer_action_required = not cropping_favored or not irrigation_system_available
```

##### Task 1 — Create the Variables

##### Task 2 — Calculate Rainfall Difference
Rainfall Difference = Minimum Required Rainfall − Recorded Rainfall

##### Task 3 — Calculate Temperature Margin <br>
Temperature Margin = Maximum Safe Temperature − Current Temperature 

##### Task 4 — Update Rainfall <br>
Later in the day, the farm receives 8 mm of additional rainfall. <br>
Update the rainfall value using an assignment operator. 

##### Task 5 — Check if Rainfall Requirement Has Been Met <br>
Determine whether the updated rainfall is greater than or equal to the minimum rainfall requirement. <br>
Store the result in a Boolean variable. 

##### Task 6 — Check if Temperature Is Within the Safe Limit <br>
Determine whether the recorded temperature is less than or equal to the maximum safe temperature. <br>
Store the answer in a Boolean variable.

##### Task 7 — Determine if Crop Conditions Are Favorable <br>
The farm is considered to have Favorable Growing Conditions if: <br>
Rainfall requirement has been met <br>
AND <br>
Temperature is within the safe limit. <br>
Use a logical operator to evaluate the condition. <br>
Store the result in a Boolean variable. 

##### Task 8 — Determine if Farmer Action Is Required <br>
The farm manager should take action if:  <br>
Growing conditions are not favorable <br>
OR <br>
Irrigation is not available. <br>
Store the result in a Boolean variable. 

##### Task 9 — Generate the Weather Impact Report



```python
print("==========================================================")
print("WEATHER IMPACT ANALYSIS REPORT")
print("==========================================================")
print("")
print("Farm Location: ", farm_location)
print("")
print("Crop Type: ", crop_type)
print("")
print("Temperature: ", temperature_C)
print("")
print("Humidity: ", humidity)
print("")
print("Wind Speed: ", wind_speed)
print("")
print("Rainfall Difference: ", rainfall_difference)
print("")
print("Temperature Margin: ", temperature_margin)
print("")
print("Minimum Rainfall Requirement: ", min_rainfall_required)
print("")
print("Maximum Safe Temperature: ", max_safe_temp_C)
print("")
print("Rainfall Requirement Met: ", rainfall_requirement_met)
print("")
print("Temperature Within Safe Limit: ",temperature_safe)
print("")
print("Irrigation Available: ", irrigation_system_available)
print("")
print("Growing Conditions Favorable: ", cropping_favored)
print("")
print("Farmer Action Required: ", farmer_action_required)
print("")
print("==========================================================")
```
